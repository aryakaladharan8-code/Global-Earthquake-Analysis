# Exploratory Data Analysis Of Global Earthquake from 2020-2026

Exploratory Data Analysis (EDA) is the process of examining and visualizing a dataset to understand its structure, patterns, distributions, relationships, and trends. It helps identify data quality issues, detect unusual observations, and generate meaningful insights before performing further analysis.

### Steps for an Exploratory Data Analysis (EDA) Project

1. Data Collection

* Collect the earthquake dataset containing records from 2020 to 2026.

* Examine the dataset to understand the available earthquake-related variables.

2. Data Cleaning

* Identify and handle missing values.

* Check for duplicate earthquake records.

* Convert the earthquake time column into the appropriate datetime format.

* Check and verify the data types of the variables.

* Handle missing categorical and numerical values appropriately.

3. Data Understanding

* Examine the structure and dimensions of the dataset.

* Review the column names and data types.

* Use summary statistics to understand numerical variables.

* Identify categorical and numerical variables.

* Examine the unique values and frequency distributions of important categorical variables.

4. Univariate Analysis

* Analyze one variable at a time.

* Use histograms to study the distribution of numerical variables.

* Use count plots to analyze categorical variables.

* Examine the distribution of earthquake magnitude, depth, seismic stations, azimuthal gap, minimum distance, RMS, and measurement errors.

5. Bivariate Analysis

* Analyze relationships between two variables.

* Use scatter plots to examine relationships between numerical variables.

* Analyze relationships involving earthquake magnitude, depth, seismic stations, azimuthal gap, minimum distance, RMS, and measurement errors.

6. Multivariate Analysis

* Analyze relationships among three or more variables.

* Use correlation heatmaps to examine relationships among numerical variables.

* Use multivariate scatter plots to analyze magnitude, depth, seismic stations, and RMS.

* Analyze earthquake magnitude categories across different years.

7. Feature Engineering

* Create new features from the earthquake occurrence time.

* Extract month, day, hour, and weekday from the time column.

* Create depth categories such as shallow, intermediate, and deep.

8. Outlier Detection

* Identify possible outliers using box plots and the IQR method.

* Examine unusual earthquake magnitude values and other numerical variables.

* Determine whether the identified unusual observations are valid earthquake measurements or possible data errors.

9. Trend and Pattern Analysis

* Analyze changes in the number of recorded earthquakes across different years.

* Examine monthly variations in earthquake records.

* Identify important patterns and changes in earthquake activity within the dataset.

10. Documentation and Reporting

* Summarize the major findings obtained from the exploratory data analysis.

* Present the important patterns, relationships, trends, and unusual observations identified in the dataset.

* Prepare a conclusion based on the overall analysis.

* Provide suggestions and recommendations based on the findings.

* Highlight important data quality issues and limitations of the dataset.

#### Distribution of Earthquake Measurement and Error Metrics

The distributions of selected earthquake measurement and error metrics are visualized together using subplots. This makes it easier to compare their spread, concentration, and skewness across different variables.

**Question 04:** How are the earthquake measurement and error metrics distributed?


```python
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

columns = ['gap', 'dmin', 'rms', 'horizontalError', 'depthError', 'magError']

titles = [
    'Azimuthal Gap',
    'Minimum Distance',
    'RMS Error',
    'Horizontal Error',
    'Depth Error',
    'Magnitude Error'
]

for ax, column, title in zip(axes.flat, columns, titles):
    ax.hist(df[column], bins=30)
    ax.set_title(title)
    ax.set_xlabel(column)
    ax.set_ylabel('Frequency')

plt.suptitle('Distributions of Earthquake Measurement and Error Metrics', fontsize=16)
plt.tight_layout()
plt.show()
```

##### Interpretation

- **Azimuthal Gap (`gap`):** Most earthquake records have azimuthal gap values concentrated around 30–80 degrees. The frequency decreases as the gap increases, with a smaller number of observations extending beyond 150 degrees.

- **Minimum Distance (`dmin`):** Most observations have minimum-distance values below 10 degrees, with the highest concentration at lower values. A small number of observations extend to much larger distances, producing a right-skewed distribution.

- **RMS Error (`rms`):** RMS error values are mainly concentrated between approximately 0.5 and 1.0, with the highest frequency around 0.7–0.8. Only a smaller number of records have values above 1.0.

- **Horizontal Error (`horizontalError`):** Most horizontal error values are concentrated around 5–10 units, with the highest frequency near 7–8. The distribution gradually decreases at higher values, with relatively few observations above 12.

- **Depth Error (`depthError`):** Depth error is strongly concentrated at lower values, particularly around 1–5. Only a small number of observations have larger depth errors, indicating a strongly right-skewed distribution.

- **Magnitude Error (`magError`):** Magnitude error values are mostly concentrated between approximately 0.04 and 0.10, with the highest frequency around 0.06–0.08. Higher error values occur less frequently.

### Outlier Analysis

Outlier analysis is used to identify values that are unusually high or low compared with the rest of the observations. Box plots and statistical methods such as the IQR method can be used to detect possible outliers.

#### Box Plot of Earthquake Magnitude

A box plot is used to visualize the distribution of earthquake magnitudes and identify possible outliers. It shows the median, quartiles, spread of the data, and values that fall outside the usual range.

**Question 23:** Are there unusually high or low earthquake magnitude values?


```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['mag'])
plt.xlabel('Magnitude')
plt.title('Box Plot of Earthquake Magnitude')
plt.show()
```

#### Calculating the Interquartile Range (IQR)

The Interquartile Range (IQR) is calculated as the difference between the third quartile (Q3) and the first quartile (Q1). It represents the middle 50% of the data and is used to identify possible outliers.

```python
q1 = df['mag'].quantile(0.25)
q3 = df['mag'].quantile(0.75)

iqr = q3 - q1

print('Q1:', q1)
print('Q3:', q3)
print('IQR:', iqr)
```

#### Calculating the Outlier Boundaries

The lower and upper boundaries are calculated using the IQR method. Values below the lower boundary or above the upper boundary are considered possible outliers.

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print('Lower Bound:', lower_bound)
print('Upper Bound:', upper_bound)
```

#### Identifying Possible Magnitude Outliers

The magnitude values outside the calculated lower and upper boundaries are identified as possible outliers. These values will be examined before deciding whether they should be removed or retained.

```python
outliers = df[(df['mag'] < lower_bound) | (df['mag'] > upper_bound)]

print('Number of possible outliers:', len(outliers))
```

#### Examining the Magnitude Outliers

The identified possible outliers are examined to understand their magnitude values. This helps determine whether the unusual values are valid earthquake observations or potential data errors.

```python
outliers[['mag', 'magType', 'place']].head(20)
```

#### Checking the Range of Magnitude Outliers

The minimum and maximum magnitude values among the identified possible outliers are examined to understand the range of unusual earthquake magnitudes.

```python
print('Minimum outlier magnitude:', outliers['mag'].min())
print('Maximum outlier magnitude:', outliers['mag'].max())
```

##### Interpretation

The IQR method identified 406 possible outliers in the earthquake magnitude data. These values range from 6.3 to 8.8. Although these observations are statistically unusual compared with the majority of the dataset, they represent valid earthquake magnitude values and are therefore retained for further analysis.
